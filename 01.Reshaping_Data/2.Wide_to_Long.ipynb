# REVIEW OF WIDE TO LONG 

---

<br>

100 tiendas ficticias de retail en ciudades de Latinoamérica, con 16 columnas. El dataset está diseñado intencionalmente para que necesites cada parámetro de `wide_to_long()`:

Estructura de las columnas wide:
<span style='color:crimson'>**revenue_Q1, revenue_Q2, customers_Q1, customers_Q2, returns_Q3, returns_Q4...**

Esto te obliga a usar:

- `stubnames` → <span style='color:royalblue'>**'revenue', 'customers', 'returns'**</span> — tres métricas distintas que unpivotear
- `i` → <span style='color:royalblue'>**store_id**</span> — la Primary Key que identifica unívocamente cada tienda
- `j` → <span style='color:royalblue'>**quarter**</span> — el nombre que le das a la nueva columna con los sufijos Q1–Q4
- `sep` → <span style='color:royalblue'>**'_'**</span> — el separador entre el stub y el sufijo (sin esto, Pandas buscaría columnas llamadas revenueQ1 sin guión bajo)
- `suffix` → <span style='color:royalblue'>**r'\w+'**</span> — aquí está la trampa pedagógica: por default, suffix es <span style='color:crimson'>**r'\d+'**</span> (solo dígitos). Pero tus sufijos son Q1, Q2, etc., que empiezan con letra. Si no cambias `suffix`, `wide_to_long()` no va a encontrar ninguna columna que matchee y te devuelve un DataFrame vacío sin error aparente

Ese último punto es especialmente importante — es un bug silencioso que genera mucha confusión cuando uno está aprendiendo esta función.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('latam_retail_stores.csv')
df
```

<br>

---

# Práctica No. 1
<br>

```python
df1 = (pd.wide_to_long(df=df,
                 stubnames=['customers', 'returns', 'revenue'],
                 i=['store_id'],
                 j='quarter',
                 sep='_',
                 suffix=r'\w\d')
)

df1.head(10)
```

```python
df1 = (pd
    .wide_to_long(df=df,
                  stubnames=['customers', 'returns', 'revenue'],
                  i=['store_id'],
                  j='quarter',
                  sep='_',
                  suffix=r'\w\d')
    .sort_values(by='store_id')
)

df1.head(10)
```
